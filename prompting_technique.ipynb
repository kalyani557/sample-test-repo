Zero-shot learning: The AI performs a task using only instructions, without any examples.

Few-shot learning: The AI performs a task after seeing a small number of examples that demonstrate the desired pattern or behavior.

**Chain-of-Thought** is a technique where an AI solves a problem by breaking it into smaller reasoning steps instead of jumping directly to the answer.

Simple Example

Question:

If a pen costs ₹10 and you buy 3 pens, how much do you pay?

Without Chain-of-Thought:

₹30

With Chain-of-Thought:

1 pen = ₹10
3 pens = 3 × ₹10
Total = ₹30

**Self-Consistency** is a prompting technique where the AI generates multiple reasoning paths for the same problem and then selects the most common or consistent answer.


Question:

A shop sells 12 pencils. Raj buys 5 and then 2 more. How many pencils did Raj buy?

Reasoning Path 1:

5 + 2 = 7

Reasoning Path 2:

First purchase = 5
Second purchase = 2
Total = 7

Reasoning Path 3:

Add all pencils bought: 5 + 2 = 7

Final Answer:

7 pencils

**Tree-of-Thought** (ToT) is an advanced reasoning technique where the AI explores multiple possible reasoning paths (branches), evaluates them, and then chooses the best path to reach a solution.

**ReAct stands** for Reasoning + Acting.

It is a prompting technique where the AI:

Reasons about the problem.
Takes an action (such as searching, using a tool, retrieving information, or performing a task).
Observes the result.
Continues reasoning until it reaches the final answer.

# **Streaming**

Streaming is a technique where the AI sends its response token by token (or chunk by chunk) as it generates it, instead of waiting for the entire response to be completed.

User: Explain Machine Learning

AI:

Benefits of Streaming

✅ Faster perceived response time

✅ Better user experience

✅ Useful for long answers

✅ Users can start reading immediately

# **Token Counting**

Token Counting is the process of measuring how many tokens are used in the input and output of an AI model.

A token is a small piece of text. It can be:

A word
Part of a word
A punctuation mark
A number

Example

Sentence:

I love AI.

Possible tokens:

["I", "love", "AI", "."]

Total tokens = 4
