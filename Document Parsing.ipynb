# OCR Technology — What You Need to Know
What is OCR?
Optical Character Recognition (OCR) is a technology that converts images of text (scanned documents, photos, screenshots) into machine-readable, editable text.

Popular OCR Engines & Tools

Tesseract — Open-source, by Google. Most widely used



1. Google Cloud Vision API — High accuracy, cloud-based
2. Amazon Textract — Great for forms and tables
3. Adobe Acrobat OCR — Popular for PDF documents
4. EasyOCR — Python library, supports 80+ languages including Hindi

```python
import os #It allows Python to interact with your computer's files and folders.
import re #It is used to search, replace, and clean text.
import fitz # It is used to open and read PDF files.
from bs4 import BeautifulSoup #It is a library that parses HTML and extracts the readable text.

documents = []

# uploaded file prossising
for file in os.listdir():

    extension = os.path.splitext(file)[1].lower()

    if extension not in [".pdf", ".html", ".txt"]:
        continue

    text = ""

    # PDF
    if extension == ".pdf":
        pdf = fitz.open(file)

        for page in pdf:
            text += page.get_text()

        pdf.close()

    # HTML
    elif extension == ".html":
        with open(file, "r", encoding="utf-8") as f:
            soup = BeautifulSoup(f, "html.parser")
            text = soup.get_text(separator=" ")

    # TXT
    elif extension == ".txt":
        with open(file, "r", encoding="utf-8") as f:
            text = f.read()

    # Clean text
    text = re.sub(r"\s+", " ", text).strip()

    documents.append({
        "filename": file,
        "text": text
    })

# Display results
for doc in documents:
    print("=" * 60)
    print("File:", doc["filename"])
    print("Extracted Text:")
    print(doc["text"][:1000])  # First 1000 characters
    print()
```
