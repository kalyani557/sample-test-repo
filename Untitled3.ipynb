| Fixed-size Chunking                                    | Semantic Chunking                                   |
| ------------------------------------------------------ | --------------------------------------------------- |
| Splits text by a fixed number of characters or tokens. | Splits text based on meaning or natural boundaries. |
| Very simple and fast.                                  | More intelligent but slower.                        |
| May cut a sentence or paragraph in half.               | Keeps related ideas together.                       |
| Easy to implement.                                     | Requires NLP or embedding-based methods.            |
| Good for small projects.                               | Better for production-quality RAG systems.          |

# What is Overlap in Chunking?

When we split a large document into smaller **chunks**, overlap means repeating some text from the end of one chunk at the beginning of the **next chunk**.

The purpose is to preserve context so that important information isn't lost at the boundaries.

Example with overlap

Suppose:

Chunk size = 50 characters
Overlap = 15 characters


Chunk 1

**Artificial Intelligence is transforming healthcare.**

Chunk 2

...transforming healthcare.
Doctors use AI for disease prediction.

Notice that "transforming healthcare." appears in both chunks. That's the overlap.

# **why we need chunks why not we use direct the all data **

This is one of the most important questions in RAG, and interviewers ask it frequently.

Short answer

We don't use all the data directly because:

📏 The document may be too large for the model's context window.

🎯 Most questions only need a small part of the document.

⚡ Searching small pieces (chunks) is much faster.

💰 Sending the entire document every time is more expensive in terms of tokens.

# **Hierarchical Chunking**

Hierarchical chunking means splitting a document into multiple levels, from large sections to smaller sections, while preserving the document's structure.

Instead of cutting the text into fixed-size pieces, we follow its natural hierarchy.

**Example**

Suppose you have a book:

```
Document
│
├── Chapter 1
│      │
│      ├── Chunk 1
│      │      Introduction to AI
│      │
│      ├── Chunk 2
│      │      Artificial Intelligence
│      │
│      └── Chunk 3
│             Machine Learning
```

 Another Example


```
Employee Handbook

1. Leave Policy

    1.1 Sick Leave

    1.2 Casual Leave

2. Salary Policy

3. Work From Home Policy
```

````python
import fitz  # PyMuPDF


# Read the uploaded PDF


pdf = fitz.open("Artificial Intelligence Overview.pdf")

text = ""

for page in pdf:
    text += page.get_text()

pdf.close()




# 1. FIXED SIZE CHUNKING


def fixed_chunk(text, chunk_size):

    chunks = []

    for i in range(0, len(text), chunk_size):
        chunks.append(text[i:i + chunk_size])

    return chunks



# 2. FIXED SIZE + OVERLAP CHUNKING


def overlap_chunk(text, chunk_size, overlap):

    chunks = []

    start = 0

    while start < len(text):

        end = start + chunk_size

        chunks.append(text[start:end])

        start = end - overlap

    return chunks



# 3. SEMANTIC CHUNKING
# (Simple paragraph-based)


def semantic_chunk(text):

    paragraphs = text.split("\n\n")

    chunks = []

    for paragraph in paragraphs:

        paragraph = paragraph.strip()

        if paragraph:
            chunks.append(paragraph)

    return chunks



# Generate Chunks


fixed_chunks = fixed_chunk(text, 300)

overlap_chunks = overlap_chunk(text, 300, 50)

semantic_chunks = semantic_chunk(text)



# Print Fixed Chunks


print("\n")
print("=" * 60)
print("FIXED SIZE CHUNKS")
print("=" * 60)

for i, chunk in enumerate(fixed_chunks, 1):
    print(f"\nChunk {i}")
    print("-" * 40)
    print(chunk)



# Print Overlap Chunks


print("\n")
print("=" * 60)
print("OVERLAP CHUNKS")
print("=" * 60)

for i, chunk in enumerate(overlap_chunks, 1):
    print(f"\nChunk {i}")
    print("-" * 40)
    print(chunk)



# Print Semantic Chunks


print("\n")
print("=" * 60)
print("SEMANTIC CHUNKS")
print("=" * 60)

for i, chunk in enumerate(semantic_chunks, 1):
    print(f"\nChunk {i}")
    print("-" * 40)
    print(chunk)
````
