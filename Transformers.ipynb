**Prompt Templates **in LangChain are reusable prompt structures that allow you to insert variables into a prompt dynamically.

eg Explain {topic} to a beginner.


Prompt versioning means keeping track of different versions of a prompt as you improve or modify it over time.

Why Version Prompts?
Track improvements
Which prompt performs better?
Rollback if needed
If a new prompt performs worse, go back to an older version.
Team collaboration
Multiple developers can see what changed.
A/B testing
Compare two prompt versions on the same task.

```python
# ── Personas ──────────────────────────────────────
personas = {
    "arjun":  {"name":"Arjun",     "level":"beginner", "tone":"friendly",  "lang":"Python"},
    "priya":  {"name":"Dr. Priya", "level":"expert",   "tone":"technical", "lang":"Python"},
    "rahul":  {"name":"Rahul",     "level":"intermediate","tone":"casual",  "lang":"SQL"},
}

# ── Templates ─────────────────────────────────────
templates = {
    "explain": "Explain {topic} to {name} ({level} level). Tone: {tone}. Use {lang} examples.",
    "code":    "Write {lang} code for {topic}. Audience: {name} ({level}). Style: {tone}.",
    "project": "Suggest a {level}-level {lang} project about {topic} for {name}. Tone: {tone}.",
}

# ── Generate ──────────────────────────────────────
def generate(persona, template, topic):
    data = personas[persona] | {"topic": topic}   # merge topic into persona dict
    return templates[template].format(**data)      # fill placeholders

# ── Try it ────────────────────────────────────────
print(generate("arjun", "explain", "backpropagation"))
print()
print(generate("priya", "code",    "attention mechanism"))
print()
print(generate("rahul", "project", "data visualization"))
```
