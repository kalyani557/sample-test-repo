```python
text = """
Machine Learning is a subset of Artificial Intelligence.

Deep Learning is a branch of Machine Learning that uses neural networks.

Python is one of the most popular programming languages for AI development.

Vector databases store embeddings and enable semantic search.
"""
```

```python
# Chunk the document
chunks = [chunk.strip() for chunk in text.split("\n\n")]

for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}:")
    print(chunk)
    print()
```

```python
#  Convert chunks into documents
from llama_index.core import Document

documents = [Document(text=chunk) for chunk in chunks]
```

```python
# Load Embedding Model
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

embed_model = HuggingFaceEmbedding(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
# Build vector index
from llama_index.core import VectorStoreIndex
from llama_index.core import Settings

Settings.embed_model = embed_model

index = VectorStoreIndex.from_documents(
    documents
)

print("Vector Index Created Successfully")
```

```python
from llama_index.llms.huggingface import HuggingFaceLLM
from llama_index.core import Settings

# Configure LlamaIndex to use a local HuggingFace LLM
# This model is a small, efficient LLM suitable for local inference.
# You might need to install 'llama-index-llms-huggingface' and 'accelerate' if not already present.
llm = HuggingFaceLLM(
    model_name="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    tokenizer_name="TinyLlama/TinyLlama-1.1B-Chat-v1.0"
)
Settings.llm = llm

query_engine = index.as_query_engine()

response = query_engine.query(
    "Which language is popular for AI?"
)

print(response)
```

```python
# try more queries
queries = [
    "What is Deep Learning?",
    "What are vector databases?",
    "What is Machine Learning?"
]

for q in queries:
    response = query_engine.query(q)

    print(f"\nQuestion: {q}")
    print("Answer:", response)
```
