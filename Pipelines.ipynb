What is the **Generator in the Retriever + Generator** (RAG) Paradigm?

The Generator is the AI model (LLM) that creates the final answer using the information provided by the retriever.

# **What is an Indexing Pipeline?**

An Indexing Pipeline is the process of collecting data, processing it, and storing it in a searchable format so that a retriever can find it quickly later.


Real-Life Example: Library 📚

Suppose a library receives 10,000 new books.

Before readers can search for them, the librarian:

Receives the books
Categorizes them
Assigns IDs
Creates an index
Places them on shelves

# What is a Query Pipeline?

A Query Pipeline is the sequence of steps that happens after a user asks a question. Its job is to take the query, find relevant information, and produce the final answer.
