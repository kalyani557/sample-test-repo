# In LangChain,
 a Chain is a sequence of steps where the output of one step becomes the input of the next step.

Think of it like a pipeline:

User Query → Prompt → LLM → Output



```
| Chain Type                  | Purpose                                                  |
| --------------------------- | -------------------------------------------------------- |
| **LLM Chain**               | Prompt → LLM → Response                                  |
| **Sequential Chain**        | Multiple chains executed one after another               |
| **Simple Sequential Chain** | Single input/output sequence                             |
| **Retrieval Chain**         | RAG: Retriever + LLM                                     |
| **Router Chain**            | Routes queries to different chains                       |
| **Conversation Chain**      | Chat with memory/history                                 |
| **Transform Chain**         | Transform data before passing it on                      |
| **Map-Reduce Chain**        | Process large documents in chunks                        |
| **Refine Chain**            | Incrementally improve an answer using multiple documents |
| **Stuff Documents Chain**   | Put all retrieved documents into one prompt              |
| **Map-Rerank Chain**        | Process documents separately and rank answers            |

```



# **LCEL (LangChain Expression Language)**

LCEL is the modern way to build chains in LangChain. It uses the | operator to connect components together.

Memory in LangChain

Memory allows the LLM to remember previous interactions in a conversation.

Without memory:

User: My name is Yatin.
AI: Nice to meet you.

User: What is my name?
AI: I don't know.

With memory:

User: My name is Yatin.
AI: Nice to meet you.

User: What is my name?
AI: Your name is Yatin.

# LangChain vs LlamaIndex

| Feature                | LangChain                              | LlamaIndex                    |
| ---------------------- | -------------------------------------- | ----------------------------- |

| Strength               | Chains, Agents, Tools, Workflows       | Retrieval, Indexing, Querying |
| Learning Curve         | Higher                                 | Easier                        |
| RAG Support            | Good                                   | Excellent                     |
| Agent Support          | Excellent                              | Basic                         |
| Memory Support         | Strong                                 | Limited                       |
| Workflow Orchestration | Strong                                 | Moderate                      |
| Data Connectors        | Many                                   | Many                          |
| Best For               | AI Assistants, Agents, Multi-step Apps | Document Q&A, Knowledge Bases |
