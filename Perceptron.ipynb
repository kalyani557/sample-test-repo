# **A Perceptron is the simplest form of an artificial neuron. It takes some inputs, performs a calculation, and produces an output.**

**"The prediction is based on multiple factors (features)."**

A basic perceptron is usually used for binary classification, meaning it makes a Yes/No, True/False, or 0/1 decision.

```python
cloudy = 1
rain_predicted = 1

score = cloudy + rain_predicted

if score >= 2:
    decision = 1
else:
    decision = 0


print(decision)
```

**An activation function decides:**

"Should the neuron activate (fire) or not?"

```python
salary = 1
credit_score = 1

score = salary + credit_score - 1.5

if score > 0:
    output = 1  # Approve
else:
    output = 0  # Reject

print(output)
```

## Loss Function — Definition

A loss function measures how wrong a model's prediction is compared to the actual answer.

```python
actual = 90
predicted = 80

loss = abs(actual - predicted)

print("Loss:", loss)
```

# Backpropagation

The idea of backpropagation is:

**Make a prediction → see the error → figure out what caused the error → adjust and improve.**
there are some eg of the backpropagation are GPS Navigation ,Cooking Tea

# Weights are always numeric values.

```python
input_value = 5 # there we can start the code
target = 10 # the targate which is wana be possible

weight = 1. #weights

for i in range(5):

    prediction = input_value * weight

    loss = target - prediction

    weight = weight + 0.1 * loss # increasing weights after itration to achive


    print(f"Iteration {i+1}")
    print("Prediction:", prediction)
    print("Loss:", loss)
    print("Weight:", weight)
    print()
```

We usually don't need exactly 10. We stop when the prediction is extremely close to 10 and the loss is very small.
