# BM25 + Dense Embeddings is called hybrid search. It combines:

BM25 (keyword-based retrieval) → finds documents using exact word matching.
Dense embeddings (semantic retrieval) → finds documents based on meaning, even if the exact words are different.


BM25 + Dense Embeddings is a hybrid retrieval technique used in RAG systems. BM25 retrieves documents through exact keyword matching, while dense embeddings retrieve documents based on semantic meaning using vector similarity. Combining them improves both precision and recall, allowing the system to handle exact terms as well as synonyms and paraphrases. This generally produces more relevant context for the LLM than using either method alone.



```
Example

Documents

D1: "Python is a programming language."
D2: "Cats are domestic animals."
D3: "Machine learning is a branch of AI."

Query

"Python programming"

BM25 scores documents based on:

Frequency of the keywords
Importance of the keywords
Document length

It will rank:

D1 ✅ (contains both words)
D3
D2
```



When using hybrid search, BM25 and dense embeddings each produce their own ranked list of documents. Hybrid fusion is the process of combining these results into a single ranking.

The two most common methods are:

Reciprocal Rank Fusion (RRF)
Linear Combination (Weighted Score Fusion)
