# **Faithfulness**
 checks whether the answer is actually supported by the documents/context provided to the model.

Example

Retrieved context:

"The Eiffel Tower is located in Paris, France. It was completed in 1889."

Question:

"Where is the Eiffel Tower and when was it completed?"

Answer A:

"The Eiffel Tower is in Paris, France and was completed in 1889."

✅ Faithful — every claim comes from the context.

Answer B:

"The Eiffel Tower is in Paris and was completed in 1889. It was designed by Leonardo da Vinci."

❌ Not faithful — "designed by Leonardo da Vinci" is not supported

# **Answer Relevancy **
measures how well the generated answer addresses the user's question.

Unlike Faithfulness, which checks whether the answer is supported by the retrieved context, Answer Relevancy checks whether the answer is actually relevant to what was asked.

Example 1: Relevant Answer

Question:

What is the capital of France?

Answer:

Paris is the capital of France.

✅ High Answer Relevancy

Example 2: Irrelevant Answer

Question:

What is the capital of France?

Answer:

France is a country in Western Europe known for its culture and cuisine.

❌ Low Answer Relevancy

The answer is factually correct but doesn't answer the question.

# **RAGAS Context Precision**

Context Precision measures how much of the retrieved context is actually useful for answering the question.

In simple terms:

Context Precision checks whether the retriever returned relevant chunks instead of a lot of unnecessary information.

Example

Question:

Who invented the telephone?

Retrieved Chunks:

✅ Alexander Graham Bell invented the telephone.
❌ The Industrial Revolution began in Britain.

# **UDCG**
 (Utility and Distraction-aware Cumulative Gain) is a retrieval evaluation metric for RAG systems that measures both the positive contribution of useful documents and the negative impact of distracting documents on LLM answer quality. Unlike NDCG, it allows negative utility scores and is better aligned with end-to-end RAG performance.

# **NDCG **
measures retrieval ranking quality, while CCRS measures whether the citations used in a RAG answer correctly support the generated claims. Together they evaluate retrieval effectiveness and source-grounded trustworthiness.
